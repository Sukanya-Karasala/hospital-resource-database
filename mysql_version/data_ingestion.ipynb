```python
for i in range(len(op_equipment_inv)):
    try:
        insert_operation_equipment_booking(
            db = db, 
            operation_id = op_equipment_inv.loc[i, 'operation_id'], 
            equipment_inv_id=op_equipment_inv.loc[i, 'equipment_inv_id'], 
            quantity_reserved = op_equipment_inv.loc[i, 'quantity_reserved'], 
            quantity_allocated = op_equipment_inv.loc[i, 'quantity_allocated']
        )
    except Exception as e:
        print(f'{e}')
```
